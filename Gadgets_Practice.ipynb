# 2. Multi-Criteria Filtering (AND / OR Logic)

### Que-1 Filter for orders where the Product_Category is "Audio" and the Unit_Price is above $150. Sort the final list by Unit_Price descending.

```python
df[(df['Product_Category'] == 'Audio') & (df['Unit_Price'] > 150.0)
   ].sort_values(by='Unit_Price', ascending=False).head()
```

### Que-2 Find all orders from the "Berlin" store that either had a discount greater than 0% or utilized express shipping. Sort by Order_Date.

```python

df[(df['Store_Location'] == 'Berlin') &
   ((df['Is_Express_Shipping']) |
    (df['Discount_Percent'] > 0.0))
    ].head()
```

### Que-3 Filter for products that cost less than $50 and achieved a Customer Rating of 4.5 or higher. Sort by Product_Category alphabetically.

```python

df[(df['Unit_Price'] < 50) &
   (df['Customer_Rating'] >= 4.5)
   ].sort_values(by='Product_Category', ascending=True).head()
```

### Que-4 Select all orders from "Mumbai" or "Sydney" where the Payment_Method

```python
df[(df['Store_Location'].isin(['Mumbai', 'Sydney'])) &
   (df['Payment_Method'] == 'PayPal')
   ].sort_values(by=['Store_Location','Order_Date'], ascending=True).head()
```

### Que-5 Identity orders where the Return_status is "completed" but the Customer_Rating was less than or equal to 2.0. Sort by Customer_Rating ascending.

```python
df[(df['Return_Status'] == 'Completed') &
   (df['Customer_Rating'] <= 2.0)
   ].sort_values(by='Customer_Rating', ascending=True)
```

### Que-6 Filter for orders of "Ultra-wide Monitor" or "RGB Mechanical Keyboard" where the Quantity_Ordered is greather than 5. Sort by Quantity_Ordered descending.

```python
df[(df['Product_Name']).isin(['Ultra-Wide Monitor', 'RGB Mechanical Keyboard']) &
   (df['Quantity_Ordered'] > 5)
   ].sort_values(by='Quantity_Ordered', ascending=False).head()
```

### Que-7 New York Electronic Clearance: Find all rows where the location is "New York", the category is "Gaming", and a discount was applied. Sort by Discount_Percent descending.

```python
df[(df['Store_Location'] == 'New York') &
   (df['Product_Category'] == 'Gaming') &
   (df['Discount_Percent'] > 0.0)
   ].sort_values(by='Discount_Percent', ascending=False)
```

### Que-8 Extract records from "WH-1" or "WH-4" where the quantity ordered is exactly 10. Sort by Warehouse_Code and then by Order_ID.

```python
df[(df["Warehouse_Code"].isin(['WH-1','WH-4'])) &
   (df['Quantity_Ordered'] == 10)
   ].sort_values(by=['Warehouse_Code', 'Order_Date'], ascending=True).head()
```

### Que-9 Filter for "Debit Card" transactions that occured in "London" with a customer rating strictly above 3.0 sort Chronologically

```python
df[(df['Payment_Method'] == 'Debit Card') &
   (df['Store_Location'] == 'London') &
   (df['Customer_Rating'] > 3.0)
   ].sort_values(by='Customer_Rating', ascending= True).head()
```

### Que-10 Find all express shipping orders that have a Customer_rating of 2.0 or lower (or are missing a rating). Sort by Customer_rating ascending.

```python
df[ (df['Is_Express_Shipping']) &
   (df['Customer_Rating'] <= 2.0) |
   (df['Customer_Rating'].isna())
   ].sort_values(by='Customer_Rating', ascending=True).head()
```

### Que-11 Isolate "Waerables" category orders that placed at "WH-2" with a unit price between $100 and $200. Sort by unit_price.

```python
df[(df['Product_Category'] == 'Wearables') &\
    (df['Warehouse_Code'] == 'WH-2') &\
          (df['Unit_Price'].between(100, 200, inclusive='both'))
          ].sort_values(by='Unit_Price', ascending=True).head(10)
```

### Que-12 Filter for orders paid via "Crypto" that have a return_status of "No_return". Sort by Quantity_ordered from highest to lowest.

```python
df[(df['Payment_Method'] == 'Crypto') &\
    (df['Return_Status'] == 'No Return')
    ].sort_values(by='Quantity_Ordered', ascending=False).head()
```
